```python
from bs4 import BeautifulSoup as bs
import pandas as pd

with open("(1) 3D models by Ddiaz Design (@ddiaz-design) - Sketchfab.html", "r") as f:
    file = f.read()

soup = bs(file, "html.parser")
itemTitle = []
itemViews = []
itemComments = []
itemLikes = []
temp = []
itemViewsParentClass = []
itemCommentsParentClass = []
itemLikesParentClass = []
items = soup.find_all("div", class_="c-grid__item item")

for item in items:
    itemTitle.append(item.find_all("a", class_="model-name__label")[0].text)
    itemViewsParentClass.append(item.find_all("span", class_="card__footer__stats --views help")[0])
    itemCommentsParentClass.append(item.find_all("a", class_="card__footer__stats --comments help")[0])
    temp = (item.find_all("div", class_="help c-like-button c-star-like-button card__footer__stats --likes"))
    if temp: #because some values contain null so cannot append a value from null list because it can't have [0] index
        itemLikesParentClass.append(temp[0])
    else:
        itemLikesParentClass.append(None)

for i in itemViewsParentClass:
    if "k" in i.find_all("span", class_="count")[0].text:
        itemViews.append(float(i.find_all("span", class_="count")[0].text.replace("k",""))*1000)
    else:
        itemViews.append(float(i.find_all("span", class_="count")[0].text))

for i in itemCommentsParentClass:
    if "k" in itemComments.append(i.find_all("span", class_="count")[0].text):
        itemComments.append(float(i.find_all("span", class_="count")[0].text.replace("k",""))*1000)
    else:
        itemComments.append(float(i.find_all("span", class_="count")[0].text.replace("k","")))

for i in itemLikesParentClass:
    if i: #because some contain null values
        if "k" in i.find_all("span", class_="count")[0].text:
            itemLikes.append(float(i.find_all("span", class_="count")[0].text.replace("k",""))*1000)
        else:
            itemLikes.append(float(i.find_all("span", class_="count")[0].text))
    else:
        itemLikes.append([])

print(len(itemViews))
print(len(itemComments))
print(len(itemLikes))

df = pd.DataFrame({"itemTitle":itemTitle, "itemViews":itemViews, "itemComments":itemComments, "itemLikes":itemLikes})
df
```
