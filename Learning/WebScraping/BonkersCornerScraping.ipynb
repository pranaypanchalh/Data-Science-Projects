```python
from bs4 import BeautifulSoup as bs
import pandas as pd
import numpy as np
import requests as rq

arrProductTitle = np.array([])
arrProductPrice = np.array([])
arrDiscountRate = np.array([])
df = pd.DataFrame()

#get the request and save the website as html

url = "https://www.bonkerscorner.com/collections/mens-new-arrivals"
rp = rq.get(url)

#print(rp.text)

def saveTheWebsite():
    with open("BonkersCornerWebsite.html", "w") as f:
        f.write(rp.text)

saveTheWebsite()

#start scraping contents
with open("BonkersCornerWebsite.html", "r") as f:
    content = f.read()
soup = bs(content, "html.parser")


#add product names in list
productList = soup.find_all("a", class_="product-card-title")
for productTitle in productList:
    arrProductTitle = np.append(arrProductTitle, productTitle.text)


#add product prices in list
#print(arrProductTitle)
productPrices = soup.find_all("span", class_="money")
for productPrice in productPrices:
    arrProductPrice = np.append(arrProductPrice, productPrice.text)
#print(arrProductPrice)

productDiscountRates = soup.find_all("span", class_="badge onsale top left")
for productDiscountRate in productDiscountRates:
    arrDiscountRate = np.append(arrDiscountRate, productDiscountRate.text.strip())
#print(arrDiscountRate)

df["Product Name"] = arrProductTitle

print(arrDiscountRate.size)
print(arrProductTitle.size)
print(arrProductPrice[::2].size)

print(arrProductPrice[::2])
df["Product Original Price"] = arrProductPrice[::2]
df["Product Discount Price"] = arrProductPrice[1::2]
df["Product Discound Rate"] = arrDiscountRate

print(df["Product Original Price"])
```
