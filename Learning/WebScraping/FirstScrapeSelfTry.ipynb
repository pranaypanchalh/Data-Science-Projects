```python
from bs4 import BeautifulSoup as bs
import pandas as pd
import requests as rq
import numpy as np

df = pd.DataFrame()
arrCountries = np.array([])

arrDetails = np.array([])

#get the data from website and save it as html
def fetchDataAndStore():
    url = "https://www.scrapethissite.com/pages/simple/"

    rp = rq.get(url)

    with open("Simple.html", "w") as f:
        f.write(rp.text)

#import into soup
#fetch all country names

with open("Simple.html", "r") as f:
    content = f.read()
soup = bs(content, "html.parser")
h3s = soup.find_all("h3")
for h3 in h3s:
    arrCountries = np.append(arrCountries, [h3.text.strip()])
df["Country"] = arrCountries

#fetch informations
spans = soup.find_all("span")
for span in spans:
    arrDetails = np.append(arrDetails, [span.text.strip()])

df["Capital"] = arrDetails[0::3]
df["Population"] = arrDetails[1::3]
df["Area"] = arrDetails[2::3]

df.to_csv("Scraped_Data.csv")
```
