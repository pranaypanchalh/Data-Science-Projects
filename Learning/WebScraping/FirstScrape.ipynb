```python
import requests as rq
from bs4 import BeautifulSoup as bs
import pandas as pd

url = "https://www.scrapethissite.com/pages/simple/"

rp = rq.get(url)
df = pd.DataFrame()
#print(rp.text)

def saveFileFirst():
    with open("Learning/WebScraping/Simple.html", "w+", encoding="utf-8") as f:
        f.write(rp.text)

#saveFileFirst()

with open("Simple.html") as f:
    content = f.read()







soup = bs(content, 'html.parser')
#print(soup)
h3s = soup.find_all("h3")
countries = [h3.text.strip() for h3 in h3s]
df["Country"] = countries

print(df)
detailsList = []

spans = soup.find_all("span")
for span in spans:
    detailsList.append(span.text.strip().split("/n")[0])

capital = detailsList[0::3]
populatiom = detailsList[1::3]
area = detailsList[2::3]

df["Capital"] = capital
df["Populatiom"] = populatiom
df["Area"] = area

df.to_csv("Collected_Data.csv")
```
