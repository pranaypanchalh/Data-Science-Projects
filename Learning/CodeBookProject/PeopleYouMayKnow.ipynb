```python
import json

def load_data(filename):
    with open("CodeBookMassiveData.json", "r") as f:
        return json.load(f)

def findPeopleYouMayKnow(data, id):
    priorityDict = {}
    
    for i in range(len(data['users'])):
        priorityDict[data['users'][i]['name']] = 0
        
    for firstFriend in data['users'][id-1]['friends']:
        for secondFriend in data['users'][firstFriend-1]['friends']:
            priorityDict[data['users'][secondFriend-1]['name']] += 1
            
    priorityDict = dict(sorted(priorityDict.items(), key=lambda x: x[1], reverse=True))

    priorityDict = {k:v for k,v in priorityDict.items() if v!=0}
    print(priorityDict)
    
data = load_data("CodeBookData.json")
findPeopleYouMayKnow(data, 1)
```
