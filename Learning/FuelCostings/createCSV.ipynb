```python
import pandas as pd
import numpy as np
cost = [0]
distance = []
date = []
with open("Hunter 350 Fuel Costings.txt", "r") as f:
    text = f.read()


def replaceDate(date):
    data = {"Jan":1,"Feb":2,"Mar":3,"April":4,"May":5,"June":6,"July":7,"Aug":8,"Sept":9,"Oct":10,"Nov":11,"Dec":12}
    for month, value in data.items():
        date = date.replace(month,f"-{value}-")
    return date

def crossMultiplication(a,b,c):
    return round((b*c)/a)

for data in text.split("\n"):
    cost.append(int(data.replace("₹","").split(" ")[0]))
    distance.append(data.replace("?","").replace("km","").split(" ")[1])
    date.append(replaceDate(data.split(" ")[2]))
cost = cost[:-1]


sumCost = []
sum = 0
for oneCost in cost:
    sum += oneCost
    sumCost.append(sum)
print(sumCost)
print(len(sumCost))
print(len(distance))
print(len(date))

for index, i in enumerate(distance):
    if i == "" and distance[index-1] != "":
        distance[index] = crossMultiplication(int(sumCost[index-1]), int(distance[index-1]), int(sumCost[index]))
    else:
        distance[index] = crossMultiplication(int(sumCost[index+1]), int(distance[index+1]), int(sumCost[index]))

df = pd.DataFrame({"cost":sumCost,"distance":distance,"date":date})
df.to_csv("fuel_log.csv", index=False)
df
```
