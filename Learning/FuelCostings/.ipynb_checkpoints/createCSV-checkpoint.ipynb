```python
import pandas as pd
import numpy as np
cost = []
distance = []
date = []
with open("Hunter 350 Fuel Costings.txt", "r") as f:
    text = f.read()
print(text)
def replaceDate(date):
    data = {"Jan":1,"Feb":2,"Mar":3,"April":4,"May":5,"June":6,"July":7,"Aug":8,"Sept":9,"Oct":10,"Nov":11,"Dec":12}
    for month, value in data.items():
        date = date.replace(month,f"-{value}-")
    return date

for data in text.split("\n"):
    cost.append(int(data.replace("₹","").split(" ")[0]))
    distance.append(int(data.replace("?","0").replace("km","").split(" ")[1]))
    date.append(replaceDate(data.split(" ")[2]))
sumList = []
sum = 0
for oneCost in cost:
    sum += oneCost
    sumList.append(sum)

df = pd.DataFrame({"cost":sum,"distance":distance,"date":date})
df.to_csv("fuel_log.csv", index=False)
```
