```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression

model = LinearRegression()

data = pd.read_csv("fuel_log.csv")
x = data[["distance"]]
y = data["cost"]
model.fit(x, y)
arrPred = model.predict(data[["distance"]])
arrOriginal = data["cost"].to_numpy()
print(arrPred-arrOriginal)

model.predict([[120000]])
```
