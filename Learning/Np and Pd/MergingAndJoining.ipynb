```python
import pandas as pd

employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})

departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]
})
employees
```

```python
departments
```

```python
pd.merge(employees, departments, on="DeptID") #inner join
```

```python
pd.merge(employees, departments, on="DeptID", how="left").fillna("None") #left join with fill
```

```python
pd.merge(employees, departments, on="DeptID", how="right").fillna("None") #right join with fill
```

```python
df1 = pd.DataFrame({"Name": ["Alice", "Bob"]})
df2 = pd.DataFrame({"Name": ["Charlie", "David"]})
 
pd.concat([df1, df2], axis=1)
```
